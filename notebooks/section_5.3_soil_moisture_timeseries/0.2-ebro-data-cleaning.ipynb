## Eliminación de Outliers

**Estrategia de detección y eliminación de outliers:**

Para garantizar la calidad de los datos y evitar que valores atípicos afecten el análisis y modelado predictivo,
aplicamos el **método IQR (Interquartile Range)**:

### Método IQR (Rango Intercuartílico)

1. **Cálculo de cuartiles:**
   - Q1 (primer cuartil): percentil 25
   - Q3 (tercer cuartil): percentil 75
   - IQR = Q3 - Q1

2. **Definición de límites:**
   - Límite inferior: Q1 - 1.5 × IQR
   - Límite superior: Q3 + 1.5 × IQR

3. **Tratamiento de outliers:**
   - Los valores fuera de estos límites se reemplazan por `NaN`
   - Esto preserva la integridad de los datos (valores válidos dentro del rango se mantienen)
   - Los valores NaN permiten análisis posteriores sin introducir sesgos artificiales

### Variables analizadas

La función se aplica a las siguientes variables climáticas continuas:
- **Temperatura:** `tav`, `tmax`, `tmin`
- **Humedad relativa:** `Hrav`, `Hrmax`, `Hrmin`
- **Viento:** `wv`, `wvmax`
- **Radiación solar:** `SolarRad`
- **Evapotranspiración:** `et0`
- **Precipitación:** `p`
- **Caudales:** `qmax`, `qav`, `qmin`

**Nota:** Las variables temporales (`Jdate`, `Year`, `Month`, `Day`) no se procesan ya que no son métricas continuas susceptibles a outliers.

2. **Anomalous discharge values**: Some basins (especially Sangüesa) have erroneous values ~10^9 m³/s

These transformations are applied before saving the clean data.

```python
# ===============================================
# DATA CLEANING TRANSFORMATIONS
# ===============================================

print("Aplicando transformaciones de limpieza...")

# 1. TRANSFORMAR UNIDADES DE CAUDAL EN BAILÍN
# Los valores de Bailín están en unidades incorrectas (probablemente L/s en lugar de m³/s)
# Se dividen por 1000 para corregir
if "Bailin" in climate_dict:
    print("\n1. Corrigiendo unidades de caudal en Bailín (dividiendo por 1000)...")
    
    # Guardar valores originales para comparación
    bailin_qmax_before = climate_dict["Bailin"]["qmax"].describe()
    
    climate_dict["Bailin"]["qmax"] = climate_dict["Bailin"]["qmax"] / 1000
    climate_dict["Bailin"]["qav"] = climate_dict["Bailin"]["qav"] / 1000
    climate_dict["Bailin"]["qmin"] = climate_dict["Bailin"]["qmin"] / 1000
    
    print(f"   ✓ Bailín qmax range: {climate_dict['Bailin']['qmax'].min():.4f} - {climate_dict['Bailin']['qmax'].max():.4f} m³/s")
    print(f"   ✓ Bailín qav range: {climate_dict['Bailin']['qav'].min():.4f} - {climate_dict['Bailin']['qav'].max():.4f} m³/s")
    print(f"   ✓ Bailín qmin range: {climate_dict['Bailin']['qmin'].min():.4f} - {climate_dict['Bailin']['qmin'].max():.4f} m³/s")

# 2. ELIMINAR VALORES ANÓMALOS DE CAUDAL
# Hay valores del orden de 10^9 que son errores de medición
# Se reemplazan con NaN todos los valores de qmax, qav, qmin > 1000 m³/s

print("\n2. Eliminando valores anómalos (>1000 m³/s)...")

# Contador de valores anómalos por cuenca
anomalies_count = {}

for basin in climate_dict.keys():
    count = 0
    
    # qmax
    mask_qmax = climate_dict[basin]["qmax"] > 1000
    if mask_qmax.any():
        count += mask_qmax.sum()
        climate_dict[basin].loc[mask_qmax, "qmax"] = np.nan
    
    # qav
    mask_qav = climate_dict[basin]["qav"] > 1000
    if mask_qav.any():
        count += mask_qav.sum()
        climate_dict[basin].loc[mask_qav, "qav"] = np.nan
    
    # qmin
    mask_qmin = climate_dict[basin]["qmin"] > 1000
    if mask_qmin.any():
        count += mask_qmin.sum()
        climate_dict[basin].loc[mask_qmin, "qmin"] = np.nan
    
    if count > 0:
        anomalies_count[basin] = count

# Mostrar resumen de valores anómalos eliminados
if anomalies_count:
    print("   Valores anómalos (>1000 m³/s) reemplazados con NaN:")
    for basin, count in anomalies_count.items():
        print(f"   ✓ {basin}: {count} valores anómalos detectados y eliminados")
else:
    print("   ✓ No se encontraron valores anómalos >1000 m³/s")

print("\n✅ Transformaciones de limpieza completadas exitosamente")
```

```python
# Verificar rangos después de limpieza en las cuencas más afectadas
print("Verificación de rangos después de limpieza:\n")

basins_to_check = ["Bailin", "Sangüesa", "Anduña", "Garona"]

for basin in basins_to_check:
    if basin in climate_dict:
        print(f"{basin}:")
        print(f"  qmax: {climate_dict[basin]['qmax'].min():.4f} - {climate_dict[basin]['qmax'].max():.4f} m³/s (mean: {climate_dict[basin]['qmax'].mean():.4f})")
        print(f"  qav:  {climate_dict[basin]['qav'].min():.4f} - {climate_dict[basin]['qav'].max():.4f} m³/s (mean: {climate_dict[basin]['qav'].mean():.4f})")
        print(f"  qmin: {climate_dict[basin]['qmin'].min():.4f} - {climate_dict[basin]['qmin'].max():.4f} m³/s (mean: {climate_dict[basin]['qmin'].mean():.4f})")
        print()
```
