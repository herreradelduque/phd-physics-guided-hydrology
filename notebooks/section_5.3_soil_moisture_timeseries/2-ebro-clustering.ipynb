# Clustering Analysis

## Step 4: Clustering basins based on discharge patterns

We'll group basins based on their discharge characteristics (qmin, qav, qmax) to identify similar behavior patterns.

```python
# Additional imports for clustering
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.decomposition import PCA
import seaborn as sns
```

### 4.1 Aggregate discharge statistics per basin

Calculate summary statistics for each basin to use as features for clustering.

```python
# Aggregate discharge statistics for each basin
discharge_stats = []

for basin_name, df in join_dict.items():
    # Calculate statistics for qmin, qav, qmax
    stats = {
        'Basin': basin_name,
        'P_mean': df['ymin'].mean(),
        'P_std': df['ymin'].std(),
        'P_median': df['ymin'].median(),
        'ymin_mean': df['ymin'].mean(),
        'ymin_std': df['ymin'].std(),
        'ymin_median': df['ymin'].median(),
        'yav_mean': df['yav'].mean(),
        'yav_std': df['yav'].std(),
        'yav_median': df['yav'].median(),
        'ymax_mean': df['ymax'].mean(),
        'ymax_std': df['ymax'].std(),
        'ymac_median': df['ymax'].median(),
        'qmin_mean': df['qmin'].mean(),
        'qmin_std': df['qmin'].std(),
        'qmin_median': df['qmin'].median(),
        'qav_mean': df['qav'].mean(),
        'qav_std': df['qav'].std(),
        'qav_median': df['qav'].median(),
        'qmax_mean': df['qmax'].mean(),
        'qmax_std': df['qmax'].std(),
        'qmax_median': df['qmax'].median(),
    }
    discharge_stats.append(stats)

# Create DataFrame with basin statistics
basin_stats_df = pd.DataFrame(discharge_stats)
basin_stats_df = basin_stats_df.set_index('Basin')

print("Discharge statistics per basin:")
display(basin_stats_df)
```

### 4.2 Determine optimal number of clusters

Use the elbow method to find the optimal number of clusters.

```python
# Standardize the features for clustering
scaler = StandardScaler()
basin_stats_scaled = scaler.fit_transform(basin_stats_df)

# Elbow method to determine optimal number of clusters
inertias = []
K_range = range(2, 8)

for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(basin_stats_scaled)
    inertias.append(kmeans.inertia_)

# Plot elbow curve
plt.figure(figsize=(10, 6))
plt.plot(K_range, inertias, 'bo-')
plt.xlabel('Number of clusters (k)')
plt.ylabel('Inertia')
plt.title('Elbow Method For Optimal k')
plt.grid(True)
plt.show()

print("Inertias by k:", dict(zip(K_range, inertias)))
```

### 4.3 K-Means Clustering

Apply K-means clustering with k=3 (adjust based on elbow plot above).

```python
# Perform K-means clustering with k=3
k_optimal = 3  # Adjust this based on the elbow plot
kmeans = KMeans(n_clusters=k_optimal, random_state=42, n_init=10)
basin_stats_df['Cluster_KMeans'] = kmeans.fit_predict(basin_stats_scaled)

# Display basins by cluster
print(f"K-Means Clustering Results (k={k_optimal}):\n")
for cluster_id in range(k_optimal):
    basins_in_cluster = basin_stats_df[basin_stats_df['Cluster_KMeans'] == cluster_id].index.tolist()
    print(f"Cluster {cluster_id}: {basins_in_cluster}")
    print(f"  Number of basins: {len(basins_in_cluster)}\n")

# Show cluster statistics
cluster_summary = basin_stats_df.groupby('Cluster_KMeans')[['qmin_mean', 'qav_mean', 'qmax_mean']].mean()
print("\nCluster average discharge (mean of means):")
display(cluster_summary)
```

### 4.4 Hierarchical Clustering

Alternative approach using hierarchical clustering with dendrogram visualization.

```python
# Perform hierarchical clustering
linkage_matrix = linkage(basin_stats_scaled, method='ward')

# Plot dendrogram
plt.figure(figsize=(14, 7))
dendrogram(
    linkage_matrix,
    labels=basin_stats_df.index,
    leaf_rotation=90,
    leaf_font_size=10
)
plt.title('Hierarchical Clustering Dendrogram (Ward Method)')
plt.xlabel('Basin')
plt.ylabel('Distance')
plt.tight_layout()
plt.show()

# Cut the dendrogram to get cluster assignments
from scipy.cluster.hierarchy import fcluster
basin_stats_df['Cluster_Hierarchical'] = fcluster(linkage_matrix, t=3, criterion='maxclust')

print("\nHierarchical Clustering Results (3 clusters):\n")
for cluster_id in sorted(basin_stats_df['Cluster_Hierarchical'].unique()):
    basins_in_cluster = basin_stats_df[basin_stats_df['Cluster_Hierarchical'] == cluster_id].index.tolist()
    print(f"Cluster {cluster_id}: {basins_in_cluster}")
    print(f"  Number of basins: {len(basins_in_cluster)}\n")
```

### 4.5 Visualize Clusters with PCA

Reduce dimensions to 2D using PCA to visualize the clusters.

```python
# Perform PCA for 2D visualization
pca = PCA(n_components=2)
basin_stats_pca = pca.fit_transform(basin_stats_scaled)

# Create visualization dataframe
pca_df = pd.DataFrame(
    basin_stats_pca, 
    columns=['PC1', 'PC2'],
    index=basin_stats_df.index
)
pca_df['Cluster_KMeans'] = basin_stats_df['Cluster_KMeans']
pca_df['Cluster_Hierarchical'] = basin_stats_df['Cluster_Hierarchical']

# Plot K-Means clusters
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# K-Means
scatter1 = axes[0].scatter(
    pca_df['PC1'], 
    pca_df['PC2'], 
    c=pca_df['Cluster_KMeans'], 
    cmap='viridis', 
    s=100, 
    alpha=0.6,
    edgecolors='black'
)
for basin_name, row in pca_df.iterrows():
    axes[0].annotate(basin_name, (row['PC1'], row['PC2']), fontsize=8, ha='right')
axes[0].set_xlabel(f'PC1 ({pca.explained_variance_ratio_[0]:.2%} variance)')
axes[0].set_ylabel(f'PC2 ({pca.explained_variance_ratio_[1]:.2%} variance)')
axes[0].set_title('K-Means Clustering (PCA visualization)')
axes[0].grid(True, alpha=0.3)

# Hierarchical
scatter2 = axes[1].scatter(
    pca_df['PC1'], 
    pca_df['PC2'], 
    c=pca_df['Cluster_Hierarchical'], 
    cmap='viridis', 
    s=100, 
    alpha=0.6,
    edgecolors='black'
)
for basin_name, row in pca_df.iterrows():
    axes[1].annotate(basin_name, (row['PC1'], row['PC2']), fontsize=8, ha='right')
axes[1].set_xlabel(f'PC1 ({pca.explained_variance_ratio_[0]:.2%} variance)')
axes[1].set_ylabel(f'PC2 ({pca.explained_variance_ratio_[1]:.2%} variance)')
axes[1].set_title('Hierarchical Clustering (PCA visualization)')
axes[1].grid(True, alpha=0.3)

plt.tight_layout()

plt.savefig("2-cluster_plots/kmeans_hierarchical_pca.png", dpi=300)
plt.show()

print(f"\nTotal variance explained by 2 PCs: {pca.explained_variance_ratio_.sum():.2%}")
```
