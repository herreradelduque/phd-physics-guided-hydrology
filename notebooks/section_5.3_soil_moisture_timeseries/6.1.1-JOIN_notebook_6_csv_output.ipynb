```python
base_path = "../../data/model_results"
patern_filename = [ "complete_table_120models_20260323","complete_table_120models_20260324", "notebook_6.1_time_series_models_20260324" ]

csv_files = [ f for f in os.listdir(base_path) if f.endswith("with-outliers.csv") and any(f.startswith(p) for p in patern_filename) ]
csv_files
```

```python
dfs = []

for csv_file in csv_files:
    print(f"Processing {csv_file}...")
    full_path = os.path.join(base_path, csv_file)
    dfs.append(pd.read_csv(full_path))

df_csv = pd.concat(dfs, ignore_index=True)
```

```python
# df_csv.to_csv(f"{base_path}/models_results_full.csv")
df_csv.to_csv(f"{base_path}/models_results_full_with_outliers.csv")
```
